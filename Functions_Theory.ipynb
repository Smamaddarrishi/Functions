#Functions

**1. What is the difference between a function and a method in Python?**

- Function

  1. A block of reusable code that performs a task.

  2. Can exist independently, outside of classes.

  Example:

      def add(a, b):
        return a + b
      print(add(3, 4))  # 7

- Method

  1. A function that belongs to an object (class instance).

  2. Always takes self (or cls for class methods) as the first parameter.

  Example:
  
      s = "hello"
      print(s.upper())  # HELLO (upper is a method of str)





**2. Explain the concept of function arguments and parameters in Python.**

- Parameters:

  1. Names listed inside a function definition.

  2. Work as placeholders for data the function expects.

  Example: def greet(name): ... → name is a parameter.

- Arguments:

  1. Actual values passed when calling a function.

- Example: greet("Sam") → "Sam" is an argument.

  In short: parameters = variables in definition, arguments = actual values in call.

- Ways to define:

  1. Regular functions with def

  2. Lambda functions

  3. Inner/nested functions

- Ways to call:

  1. Positional arguments

  2. Keyword arguments

  3. Default arguments

          # Different definitions

          #Different calls

**4. What is the purpose of the `return` statement in a Python function?**

- Ends the execution of a function and sends a value back to the caller.

- Without return, Python functions return None by default.

- Helps reuse results in other parts of the program.

- Example:

              def multiply(a, b):

                return a * b
                
              result = multiply(4, 5)
              print(result)  # 20


**5. What are iterators in Python and how do they differ from iterables?**

- Iterable:

  1. An object capable of returning its members one at a time.

  2. Examples: lists, tuples, strings, ranges.

  3. You can loop over them with for.

- Iterator:

  1. An object with __next__() method.

  2. Created by calling iter() on an iterable.

  3. Produces values one by one until exhausted.

- Example:

      nums = [10, 20, 30]   # iterable
      it = iter(nums)       # iterator
      print(next(it))       # 10
      print(next(it))       # 20


**6. Explain the concept of generators in Python and how they are defined.**

- A generator is a special type of iterator that yields values lazily (on demand).

- Defined using the yield keyword inside a function.

- Can also be created with generator expressions (expr for x in iterable).

- Example:

      def countdown(n):
        while n > 0:
        yield n
        n -= 1
      for num in countdown(3):
        print(num)
      # 3, 2, 1


**7. What are the advantages of using generators over regular functions?**

- Memory efficient: Don’t store all values at once (good for large data).

- Lazy evaluation: Values generated only when needed.

- Can represent infinite sequences: Unlike lists which must be finite.

- Improves performance for pipelines and streaming data.

- Example:

      def squares(n):
        for i in range(n):
          yield i*i
      for s in squares(5):
        print(s)


**8. What is a lambda function in Python and when is it typically used?**

- A lambda is a small, anonymous function defined with lambda keyword.

- Syntax: lambda arguments: expression (no return, single line only).

- Useful for short tasks where defining a full function is unnecessary.

- Commonly used with map(), filter(), sorted().

- Example:

      nums = [5, 2, 7]
      nums_sorted = sorted(nums, key=lambda x: -x)
      print(nums_sorted)  # [7, 5, 2]


**9. Explain the purpose and usage of the `map()` function in Python.**
- Transformation tool

  1. The main purpose of map() is to apply a function to each element of an iterable (like a list, tuple, or string).

  2. Instead of writing a for loop, you simply “map” a function across all items.

- Functional programming style

  1. Encourages a cleaner, more declarative approach (saying what you want instead of how to do it).

  2. Keeps code concise and readable.

- Performance

  1. map() returns an iterator, which is lazy (values are produced only when needed).

  2. This makes it memory-efficient, especially with large datasets.

- Example:

      nums = [1, 2, 3, 4]
      doubled = map(lambda x: x*2, nums)
      print(list(doubled))  # [2, 4, 6, 8]


**10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?**

- map()

  1. Purpose: Transform every element in an iterable.

  2. Function type: Function should take one (or more) inputs and return a transformed value.

  3. Output: Returns an iterator of the same length as the input iterable.

  Example:

      nums = [1, 2, 3, 4]
      squares = map(lambda x: x*x, nums)
      print(list(squares))   # [1, 4, 9, 16]

- filter()

  1. Purpose: Select elements that satisfy a condition.

  2. Function type: Function must return True or False.

  3. Output: Returns an iterator with only the elements for which the function is True.

  Example:

      nums = [1, 2, 3, 4, 5, 6]
      evens = filter(lambda x: x % 2 == 0, nums)
      print(list(evens))  # [2, 4, 6]

- reduce()

  1. Purpose: Reduce an iterable to a single cumulative value.

  2. Function type: Function must take two inputs and return a single value.

  3. Output: Returns one value (not an iterator).

  4. Note: reduce() is not built-in; it is imported from functools.

  Example:

      from functools import reduce
      nums = [1, 2, 3, 4]
      product = reduce(lambda a, b: a * b, nums)
      print(product)  # 24
